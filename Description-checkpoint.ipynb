```python
import pandas as pd

# --- Step 1: Load CSV ---
df = pd.read_csv(r"C:\Users\HP\Documents\BANK\bankStatement.csv")

# --- Step 2: Clean column names and types ---
df.columns = df.columns.str.strip()
df['Description'] = df['Description'].fillna('').astype(str)
df['Debit'] = pd.to_numeric(df['Debit'], errors='coerce').fillna(0)
df['Credit'] = pd.to_numeric(df['Credit'], errors='coerce').fillna(0)
df['Date'] = pd.to_datetime(df['Posting Date'], errors='coerce')

# Sort by Date (optional: keep original order)
df = df.sort_values(['Date']).reset_index(drop=True)

# --- Step 3: Assign TransactionKey ---
current_key = 0
keys = []

for _, row in df.iterrows():
    if row['Debit'] != 0 or row['Credit'] != 0:
        # Main row → increment key
        current_key += 1
        keys.append(current_key)
    else:
        # Drop row → same key as last main row
        keys.append(current_key)

df['TransactionKey'] = keys

# --- Step 4: Split into Main and Drop tables ---
main_df = df[(df['Debit'] != 0) | (df['Credit'] != 0)].copy()
drop_df = df[(df['Debit'] == 0) & (df['Credit'] == 0)].copy()

# --- Step 5: Merge Drop descriptions into Main table ---
merged_descriptions = drop_df.groupby('TransactionKey')['Description']\
                             .apply(lambda x: ' '.join(x)).reset_index()

main_df = main_df.merge(merged_descriptions, on='TransactionKey', how='left', suffixes=('', '_Drop'))

# Combine descriptions
main_df['FullDescription'] = main_df['Description'] + ' ' + main_df['Description_Drop'].fillna('')
main_df['FullDescription'] = main_df['FullDescription'].str.strip()

# Drop temporary columns if needed
main_df = main_df.drop(columns=['Description_Drop'])

# --- Step 6: Save results ---
main_df.to_csv(r"C:\Users\HP\Documents\BANK\main_transactions_cleaned.csv", index=False)
drop_df.to_csv(r"C:\Users\HP\Documents\BANK\drop_rows_reference.csv", index=False)

print("Cleaning completed!")
print("Main transactions saved as: main_transactions_cleaned.csv")
print("Drop rows saved as: drop_rows_reference.csv")
```
